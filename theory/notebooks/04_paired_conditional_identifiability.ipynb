# 04 Paired Conditional Identifiability

Finite witness for [`../04_paired_conditional_identifiability.md`](../04_paired_conditional_identifiability.md). This Notebook is not a proof or empirical model result.

```python
import numpy as np
x = np.array([0, 0, 1, 1])
z_plus = x.copy()
z_minus = 1 - x
assert np.array_equal(np.bincount(z_plus), np.bincount(z_minus))
assert np.array_equal(np.bincount(x), np.array([2, 2]))
p_plus = z_plus[x == 1].mean()
p_minus = z_minus[x == 1].mean()
assert p_plus == 1.0 and p_minus == 0.0
print("THEORY_DEMO_PASS::04_paired_conditional_identifiability")
```
