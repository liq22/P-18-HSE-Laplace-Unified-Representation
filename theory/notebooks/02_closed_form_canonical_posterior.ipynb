# 02 Closed Form Canonical Posterior

Finite witness for [`../02_closed_form_canonical_posterior.md`](../02_closed_form_canonical_posterior.md). This Notebook is not a proof or empirical model result.

```python
import numpy as np
from hse_laplace import gaussian_information_statistics, gaussian_posterior
A = np.array([[1., 0.], [0., 2.]])
x = np.array([0.4, -1.0])
stats = gaussian_information_statistics(A, np.eye(2), x)
posterior = gaussian_posterior(np.zeros(2), np.eye(2), stats)
expected_cov = np.diag([0.5, 0.2])
expected_mean = expected_cov @ stats.score
assert np.allclose(posterior.covariance, expected_cov)
assert np.allclose(posterior.mean, expected_mean)
print("THEORY_DEMO_PASS::02_closed_form_canonical_posterior")
```
