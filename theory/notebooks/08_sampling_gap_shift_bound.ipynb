# 08 Sampling Gap Shift Bound

Finite witness for [`../08_sampling_gap_shift_bound.md`](../08_sampling_gap_shift_bound.md). This Notebook is not a proof or empirical model result.

```python
import numpy as np
s = -0.4 + 3.0j
p = np.array([0.10, 0.20, 0.30])
q = np.array([0.12, 0.22, 0.32])
lambda_p = np.mean(np.exp(s * p))
lambda_q = np.mean(np.exp(s * q))
# Index pairing is optimal for these equally weighted sorted one-dimensional samples.
w1 = np.mean(np.abs(p - q))
assert abs(lambda_p - lambda_q) <= abs(s) * w1 + 1e-12
print("THEORY_DEMO_PASS::08_sampling_gap_shift_bound")
```
