# Independent calibration for fixed policies
Proof: [policy_certificate.md](policy_certificate.md). This is a finite witness of a standard paired Hoeffding bound, not learned-model or novelty evidence.

```python
import numpy as np
from experiments.p19.policy_certificate import certify
from experiments.p19.certified_selection_demo import experiment
small=certify(np.full(64,.3),np.full((64,3),.2),np.zeros(3),margin=.01)
large=certify(np.full(2048,.3),np.full((2048,3),.2),np.zeros(3),margin=.01)
assert small['selected_index']==-1 and large['selected_index']==0
print('n=64 radius:',small['radius'],'n=2048 radius:',large['radius'])
```

```python
records,_=experiment(2048,0)
a,b=records[2:]
np.testing.assert_allclose(a['best_lower_bound']-b['best_lower_bound'],.4)
assert a['selected_policy']=='hard_source_route' and a['test_net_improvement']<0
assert b['selected_policy']=='static_reference' and b['test_net_improvement']==0
for row in records:print(row)
```

The last two settings share observations and fitted policies. Only the declared shift allowance changes. The source-only certificate cannot detect arbitrary target shift. Unbounded Energy Score and nonlinear macro-F1 are not passed into this rule.
